```python
df["timestamp"].head(5)
```

```python
df["timestamp"]=pd.to_datetime(df["timestamp"])
```

Split it into Year,month,days,hour,min

```python
df["year"]=df["timestamp"].dt.year
df["month"]=df["timestamp"].dt.month
df["day"]=df["timestamp"].dt.day
df["hour"]=df["timestamp"].dt.hour
df["minute"]=df["timestamp"].dt.minute
```

```python
df
```

drop timestamp because we have taken all the necessary details from it

```python
df.drop(columns=["timestamp"],inplace=True)
```

```python
df
```

Map function is used here to feature encoding

```python
df["junction_id"]=df["junction_id"].map({"J1":1,"J2":2,"J3":3,"J4":4,"J5":5})
```

```python
df["junction_id"].value_counts()
```

We can Compare IQR Values(Min,Max,25%,50%,75%)

```python
sns.boxplot(df,x=df["vehicle_count"])
```

```python
sns.boxplot(df,x=df["avg_speed"])
```

Total Number of vehicles in each Junction

```python
junction_traffic=df.groupby("junction_id")["vehicle_count"].sum()
print(junction_traffic)
```

so from this we get know that Junction 4 has the highest Traffic

```python
print(junction_traffic.idxmax(),"has the highest Traffic")
```

Peak Time

```python
peak_time=df.groupby("hour")["vehicle_count"].sum()
print(peak_time)
```

```python
print(peak_time.idxmax(),"is the peakest hour")
```

```python
sns.lineplot(x=peak_time.index,y=peak_time.values)
plt.show()
```

Traffic in each junction in each hour

```python
sns.lineplot(df,x=df["hour"],y=df["vehicle_count"],hue="junction_id")
plt.title("Traffic in each Junction in each hour")
plt.show()
```

```python
peak_per_junction=df.loc[df.groupby("junction_id")["vehicle_count"].idxmax()]
print(peak_per_junction)
```
